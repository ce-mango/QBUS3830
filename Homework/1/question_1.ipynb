```python
import statsmodels.api as sm
import numpy as np
```

```python
df = sm.datasets.randhie.load_pandas().data
df = sm.add_constant(df, prepend=True)

df
```

```python
def j(B, df=df):
    rolling_sum = 0

    for i in range(len(df)):
        x = df.iloc[i].values
        y = df.iloc[i]
        rolling_sum += np.exp(x @ B) - y * (x @ B)
    return rolling_sum

B = np.array([1, 0, 0, 0, 0, 0, 0, 0.02, 0, 0, 0])

j = j(B)
j
```

```python
def grad_j(B, df=df):
    rolling_sum = np.zeros(len(B))

    for i in range(len(df)):
        x = df.iloc[i].values
        y = df.iloc[i]
        rolling_sum += (np.exp(x @ B) - y) * x
    return rolling_sum

grad_j = grad_j(B)
grad_j
```

```python
def H(B, df=df):
    return df.T @ np.diag(np.exp(df @ B)) @ df

H = H(B)
print("Intercept-intercept: ", H["const"]["const"])
```
